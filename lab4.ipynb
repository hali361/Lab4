## Test 2 - Team class

This class creates a Red Bull Racing team and adds 2 drivers

Expected result: 575 + 285 = a total of 860 points

```python
team1 = Team("Red Bull Racing Honda RBPT")

team1.add_driver(Driver("Max Verstappen", 575))
team1.add_driver(Driver("Sergio Perez", 285))

print(team1)
print("Total points:", team1.get_total_points())
```

## Test 3 - Sorting teams

This test creates 3 teams with different total points and sorts them from lowest points to highest

Expected result: Team B (50), Team C (100), Team A (200)

```python
team_a = Team("Team A")
team_a.add_driver(Driver("Driver A", 200))

team_b = Team("Team B")
team_b.add_driver(Driver("Driver B", 50))

team_c = Team("Team C")
team_c.add_driver(Driver("Driver C", 100))

teams = [team_a, team_b, team_c]

for team in sorted(teams):
    print(team.name, team.get_total_points())
```

## Test 6 - Missing CSV file

This test checks if the program handles a CSV file that is missing without crashing

Input: Runs the program in a temporary folder without f1_points.csv

Expected output: Error: f1_points.csv file was not found

```python
import subprocess
import sys
import tempfile
from pathlib import Path

with tempfile.TemporaryDirectory() as test_folder:
    main_file = Path("main3.py").resolve()

    result = subprocess.run(
        [sys.executable, str(main_file)],
        capture_output=True,
        text=True,
        cwd=test_folder,
    )

    print("Expected: Error: f1_points.csv was not found.")
    print("Actual:", result.stdout.strip())
```

## Test 7 - Invalid points data

This test checks if the program handles invalid points within the CSV file

Input: A CSV file which contains "abc" instead of a number

Expected output: Error: Invalid points value in f1_points.csv

```python
with tempfile.TemporaryDirectory() as test_folder:
    test_csv = Path(test_folder) / "f1_points.csv"

    test_csv.write_text(
        "Driver,Team,Points\n"
        "Test driver,Test team,abc\n",
        encoding="utf-8"
    )

    result = subprocess.run(
        [sys.executable, str(Path("main3.py").resolve())],
        capture_output=True,
        text=True,
        cwd=test_folder,
    )

    print("Expected: Error: Invalid points value in f1_points.csv.")
    print("Actual:", result.stdout.strip())
```
